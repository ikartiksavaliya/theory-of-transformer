## Bag-of-Words Approach (Language Representation)

- For models to process **language**, we convert text and **sentences** into numerical **vector representation**
- In **Bag-of-Words approach**:
  - Input **sentence** → Split input by whitespace → **Tokens**
  - Build **vocabulary** from unique words
  - Count individual words in the **sentence** → **Vector representation**
